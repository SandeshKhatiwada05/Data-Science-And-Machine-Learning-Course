```python
import numpy as np
import torch
from datetime import datetime
```

```python
d = 30000
```

```python
#numpy
start_numpy = datetime.now()
A = np.random.rand(d, d).astype(np.float32)
B = np.random.rand(d, d).astype(np.float32)
C = A.dot(B)
end_numpy = datetime.now()
print(end_numpy-start_numpy)
```

```python
#using torch with gpu
start_torch = datetime.now()
A = torch.rand(d, d)
B = torch.rand(d, d)
C = torch.mm(A, B)
end_torch = datetime.now()
print(end_torch-start_torch)
```

```python
print("Total difference here", ((end_torch-start_torch) - (end_numpy-start_numpy)))
```
